## 3. Passo de Pesquisa – `find_references`

No laboratório avaliado, a função que você implementou tanto **pesquisava na web quanto escrevia um rascunho** em um único passo.

Aqui, separamos a funcionalidade de busca na web em uma função separada chamada `find_references`. Isso permite avaliar os resultados da busca independentemente dos passos de escrita e reflexão, que deixaremos de fora deste laboratório já que estamos focando apenas na saída do passo de busca na web.

Observe duas diferenças principais da implementação do laboratório avaliado:

* Esta nova função usa **AISuite**, que gerencia automaticamente as chamadas de ferramentas para você (em vez de escrever código de chamada de ferramenta manual com o SDK da OpenAI).  
* A função também informa ao LLM a **data atual**, o que ajuda a melhorar a relevância para consultas sensíveis ao tempo.  

O papel de `find_references` é **coletar informações externas** de ferramentas como **Arxiv**, **Tavily** e **Wikipedia**.  
Como a qualidade desses resultados molda diretamente as saídas do laboratório avaliado, este é o estágio onde você pode aplicar **métodos de avaliação** — por exemplo, verificando se as URLs retornadas vêm da sua lista de **domínios preferenciais**.  

```python
def find_references(task: str, model: str = "openai:gpt-4o", return_messages: bool = False):
    """Perform a research task using external tools (arxiv, tavily, wikipedia)."""

    prompt = f"""
    You are a research function with access to:
    - arxiv_tool: academic papers
    - tavily_tool: general web search (return JSON when asked)
    - wikipedia_tool: encyclopedic summaries

    Task:
    {task}

    Today is {datetime.now().strftime('%Y-%m-%d')}.
    """.strip()

    messages = [{"role": "user", "content": prompt}]
    tools = [
        research_tools.arxiv_search_tool,
        research_tools.tavily_search_tool,
        research_tools.wikipedia_search_tool,
    ]

    try:
        response = client.chat.completions.create(
            model=model,
            messages=messages,
            tools=tools,
            tool_choice="auto",
            max_turns=5,
        )
        content = response.choices[0].message.content
        return (content, messages) if return_messages else content
    except Exception as e:
        return f"[Model Error: {e}]"
```

Execute a célula a seguir para testar a função de pesquisa.

```python
research_task = "Find 2 recent papers about recent developments in black hole science"
research_result = find_references(research_task)

utils.print_html(
    research_result,
    title="Research Function Output"
)
```

## 4. Passo de Avaliação – Domínios Preferenciais

Nem todas as fontes recuperadas pela busca na web são igualmente confiáveis.  
Neste laboratório, focamos em **apenas um passo do laboratório avaliado anterior** — o passo de pesquisa `find_references` — e mostramos como projetar uma **avaliação em nível de componente** que verifica se os domínios retornados pertencem a uma lista predefinida de **domínios preferenciais**.  

Este é um exemplo de uma **avaliação objetiva com um 'ground truth' claro por exemplo**.  
Como lembrete da aula, recorde os dois eixos de avaliação: ao longo desses eixos estamos trabalhando no **quadrante superior esquerdo** — avaliações objetivas com 'ground truth' explicitamente definido aplicado no nível de cada exemplo.

<img src='M4-UGL-1-Evaluations.png' width='80%'>


### Por que avaliações em nível de componente?

Como Andrew mencionou na aula:  

- Se o problema está na busca web (geralmente o **primeiro passo** em um fluxo de trabalho de laboratório avaliado), reexecutar todo o pipeline (busca → rascunho → reflexão) a cada vez pode ser **caro** e ruidoso.  
- Pequenas melhorias na qualidade da busca web podem ser escondidas pela aleatoriedade introduzida por componentes posteriores.  
- Avaliando a busca na web *isoladamente*, você obtém um **sinal mais claro** de se esse componente está melhorando.  

Avaliações em nível de componente também são eficientes quando múltiplas equipes estão trabalhando em diferentes peças de um sistema: cada equipe pode otimizar seu próprio componente usando uma métrica clara, sem precisar rodar ou esperar por testes ponta a ponta completos.  

### Como avaliamos?

Nossa avaliação aqui é **objetiva**, e portanto pode ser avaliada usando código. Ela tem um 'ground truth' específico do exemplo - a lista de fontes preferenciais para esta consulta sobre buracos negros. Para construir a avaliação, você irá:

1. Extrair as URLs retornadas pelo Tavily.  
2. Compará-las com uma lista predefinida de **domínios preferenciais** (ex: `arxiv.org`, `nature.com`, `nasa.gov`).  
3. Calcular a **proporção de resultados preferenciais vs. total**.  
4. Retornar uma **flag PASS/FAIL** junto com um resumo formatado em Markdown.  

Isso fornece uma métrica reproduzível e de baixo custo que nos diz se o componente de pesquisa — e apenas este passo do laboratório avaliado — está buscando de fontes confiáveis.



```python
# list of preferred domains for Tavily results
TOP_DOMAINS = {
    # General reference / institutions / publishers
    "wikipedia.org", "nature.com", "science.org", "sciencemag.org", "cell.com",
    "mit.edu", "stanford.edu", "harvard.edu", "nasa.gov", "noaa.gov", "europa.eu",

    # CS/AI venues & indexes
    "arxiv.org", "acm.org", "ieee.org", "neurips.cc", "icml.cc", "openreview.net",

    # Other reputable outlets
    "elifesciences.org", "pnas.org", "jmlr.org", "springer.com", "sciencedirect.com",

    # Extra domains (case-specific additions)
    "pbs.org", "nova.edu", "nvcc.edu", "cccco.edu",

    # Well known programming sites
    "codecademy.com", "datacamp.com"
}

def evaluate_tavily_results(TOP_DOMAINS, raw: str, min_ratio=0.4):
    """
    Evaluate whether plain-text research results mostly come from preferred domains.

    Args:
        TOP_DOMAINS (set[str]): Set of preferred domains (e.g., 'arxiv.org', 'nature.com').
        raw (str): Plain text or Markdown containing URLs.
        min_ratio (float): Minimum preferred ratio required to pass (e.g., 0.4 = 40%).

    Returns:
        tuple[bool, str]: (flag, markdown_report)
            flag -> True if PASS, False if FAIL
            markdown_report -> Markdown-formatted summary of the evaluation
    """

    # Extract URLs from the text
    url_pattern = re.compile(r'https?://[^\s\]\)>\}]+', flags=re.IGNORECASE)
    urls = url_pattern.findall(raw)

    if not urls:
        return False, """### Evaluation — Tavily Preferred Domains
No URLs detected in the provided text. 
Please include links in your research results.
"""

    # Count preferred vs total
    total = len(urls)
    preferred_count = 0
    details = []

    for url in urls:
        domain = url.split("/")[2]
        preferred = any(td in domain for td in TOP_DOMAINS)
        if preferred:
            preferred_count += 1
        details.append(f"- {url} → {'✅ PREFERRED' if preferred else '❌ NOT PREFERRED'}")

    ratio = preferred_count / total if total > 0 else 0.0
    flag = ratio >= min_ratio

    # Markdown report
    report = f"""
### Evaluation — Tavily Preferred Domains
- Total results: {total}
- Preferred results: {preferred_count}
- Ratio: {ratio:.2%}
- Threshold: {min_ratio:.0%}
- Status: {"✅ PASS" if flag else "❌ FAIL"}

**Details:**
{chr(10).join(details)}
"""
    return flag, report
```

<strong>🔎 Por que esta é uma avaliação objetiva:</strong><br><br>  
Cada URL recuperada do Tavily é comparada com uma lista predefinida de <em>domínios preferenciais</em> (<code>TOP_DOMAINS</code>):<br>  
• Se o domínio corresponde → ✅ PREFERRED<br>  
• Caso contrário → ❌ NOT PREFERRED<br><br>  
Isso fornece um sinal claro de PASS/FAIL dependendo se a proporção de fontes preferenciais excede um determinado limite.  
Como o 'ground truth' (preferencial vs. não preferencial) é explicitamente definido para cada exemplo, a avaliação é tanto <strong>objetiva</strong> quanto <strong>reproduzível</strong>.
